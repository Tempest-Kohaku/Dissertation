```python
import os
import random
import json
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm
import torch
import torch.nn as nn
from torchvision import models, transforms
import shap
import matplotlib.pyplot as plt
import cv2
from scipy.stats import spearmanr

# Configuration
SEED = 42
IMG_SIZE = 224
BATCH_SIZE = 8
BACKGROUND_SIZE = 50
NSAMPLES = 200
MODEL_PATH = "outputs/models/best_densenet121_tb.pth"
TRAIN_CSV = "outputs/train_split.csv"
SHAP_VALUES_PATH = "outputs/shap/shap_values.npy"
SHAP_METADATA_PATH = "outputs/shap/shap_metadata.csv"
OUTPUT_DIR = "outputs/stability"
PLOT_DIR = os.path.join(OUTPUT_DIR, "plots")
EXAMPLE_DIR = os.path.join(OUTPUT_DIR, "example_stability_panels")
PERT_SHAP_DIR = os.path.join(OUTPUT_DIR, "perturbed_shap_values")

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(PLOT_DIR, exist_ok=True)
os.makedirs(EXAMPLE_DIR, exist_ok=True)
os.makedirs(PERT_SHAP_DIR, exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", DEVICE)

# Reproducibility
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
set_seed(SEED)

# ImageNet normalisation constants
IMAGENET_MEAN = np.array([0.485, 0.456, 0.406])
IMAGENET_STD = np.array([0.229, 0.224, 0.225])

# Dataset for SHAP background
class TBXrayDataset(torch.utils.data.Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        image_path = row["image_path"]
        label = int(row["label"])
        image_id = row["image_id"]
        image = Image.open(image_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, label, image_id, image_path

# Transforms
base_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=IMAGENET_MEAN.tolist(),
        std=IMAGENET_STD.tolist()
    )
])

# Load DenseNet121 model
model = models.densenet121(weights=None)
num_features = model.classifier.in_features
model.classifier = nn.Linear(num_features, 2)
model.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE))
model = model.to(DEVICE)
model.eval()
```

```python
# SHAP wrapper
class TBLogitModel(nn.Module):
    """
    Return only the TB class logit for SHAP.
    """
    def __init__(self, base_model):
        super().__init__()
        self.base_model = base_model

    def forward(self, x):
        outputs = self.base_model(x)
        return outputs[:, 1:2]
        
wrapped_model = TBLogitModel(model).to(DEVICE)
wrapped_model.eval()

# Helper functions
def load_image_as_display_np(image_path):
    """
    Load image, resize to 224x224, return numpy image in [0,1].
    """
    image = Image.open(image_path).convert("RGB")
    image = image.resize((IMG_SIZE, IMG_SIZE))
    image_np = np.array(image).astype(np.float32) / 255.0
    return image_np

def display_np_to_normalized_tensor(image_np):
    """
    Convert display image [0,1], H x W x 3 into ImageNet-normalised tensor.
    """
    image_np = np.clip(image_np, 0, 1)
    normalized = (image_np - IMAGENET_MEAN) / IMAGENET_STD
    tensor = torch.tensor(normalized.transpose(2, 0, 1), dtype=torch.float32)
    return tensor

def tensor_to_display_np(tensor):
    """
    Convert ImageNet-normalised tensor to display image [0,1].
    """
    image = tensor.detach().cpu().numpy().transpose(1, 2, 0)
    image = IMAGENET_STD * image + IMAGENET_MEAN
    image = np.clip(image, 0, 1)
    return image

def shap_to_heatmap(shap_value):
    """
    Convert SHAP values to a normalised 2D heatmap.
    Expected input: C x H x W or H x W x C
    """
    shap_value = np.asarray(shap_value)
    if shap_value.ndim != 3:
        raise ValueError(f"Expected 3D SHAP value, got shape {shap_value.shape}")

    if shap_value.shape[0] == 3:
        heatmap = np.abs(shap_value).mean(axis=0)
    elif shap_value.shape[-1] == 3:
        heatmap = np.abs(shap_value).mean(axis=-1)
    else:
        raise ValueError(f"Unexpected SHAP shape: {shap_value.shape}")

    heatmap = heatmap - heatmap.min()
    heatmap = heatmap / (heatmap.max() + 1e-8)
    return heatmap

def normalise_shap_return(shap_values):
    """
    Convert SHAP output into N x C x H x W.
    """
    if isinstance(shap_values, list):
        shap_values = shap_values[0]

    shap_values = np.asarray(shap_values)

    if shap_values.ndim == 5 and shap_values.shape[-1] == 1:
        shap_values = shap_values[..., 0]

    if shap_values.ndim == 4 and shap_values.shape[-1] == 3:
        shap_values = np.transpose(shap_values, (0, 3, 1, 2))

    if shap_values.ndim != 4:
        raise ValueError(f"Unexpected SHAP values shape: {shap_values.shape}")

    if shap_values.shape[1] != 3:
        raise ValueError(f"Expected N x C x H x W, got {shap_values.shape}")

    return shap_values

def get_tb_probability(model, tensor):
    """
    Get DenseNet121 TB probability and predicted class.
    """
    model.eval()

    with torch.no_grad():
        tensor = tensor.unsqueeze(0).to(DEVICE)
        logits = model(tensor)
        probs = torch.softmax(logits, dim=1)

        tb_prob = probs[0, 1].item()
        pred = torch.argmax(probs, dim=1).item()

    return tb_prob, pred

def perturb_image(image_np, perturbation_type):
    """
    Apply small perturbation to image_np in [0,1].
    """
    image_np = np.clip(image_np, 0, 1)

    if perturbation_type == "brightness_plus":
        perturbed = image_np * 1.10

    elif perturbation_type == "brightness_minus":
        perturbed = image_np * 0.90

    elif perturbation_type == "contrast_plus":
        mean = image_np.mean(axis=(0, 1), keepdims=True)
        perturbed = (image_np - mean) * 1.10 + mean

    elif perturbation_type == "contrast_minus":
        mean = image_np.mean(axis=(0, 1), keepdims=True)
        perturbed = (image_np - mean) * 0.90 + mean

    elif perturbation_type == "gaussian_noise":
        noise = np.random.normal(0, 0.02, image_np.shape)
        perturbed = image_np + noise

    elif perturbation_type == "rotation":
        h, w = image_np.shape[:2]
        matrix = cv2.getRotationMatrix2D((w // 2, h // 2), 3, 1.0)
        perturbed = cv2.warpAffine(
            image_np,
            matrix,
            (w, h),
            flags=cv2.INTER_LINEAR,
            borderMode=cv2.BORDER_REFLECT
        )

    else:
        raise ValueError(f"Unknown perturbation type: {perturbation_type}")

    return np.clip(perturbed, 0, 1)

def cosine_similarity(a, b):
    a = a.flatten()
    b = b.flatten()
    numerator = np.dot(a, b)
    denominator = np.linalg.norm(a) * np.linalg.norm(b)
    if denominator == 0:
        return 0.0
    return numerator / denominator

def spearman_similarity(a, b):
    a = a.flatten()
    b = b.flatten()
    corr, _ = spearmanr(a, b)
    if np.isnan(corr):
        return 0.0
    return corr

def jaccard_topk(a, b, top_percent=10):
    """
    Compare overlap between top-k important pixels.
    """
    a_flat = a.flatten()
    b_flat = b.flatten()
    n = len(a_flat)
    k = int((top_percent / 100.0) * n)
    a_top = set(np.argsort(a_flat)[-k:])
    b_top = set(np.argsort(b_flat)[-k:])
    intersection = len(a_top.intersection(b_top))
    union = len(a_top.union(b_top))
    if union == 0:
        return 0.0

    return intersection / union

def create_overlay(original_image, heatmap):
    heatmap_resized = cv2.resize(
        heatmap,
        (original_image.shape[1], original_image.shape[0])
    )
    heatmap_uint8 = np.uint8(255 * heatmap_resized)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB) / 255.0
    overlay = 0.55 * original_image + 0.45 * heatmap_color
    overlay = np.clip(overlay, 0, 1)

    return overlay

def save_stability_panel(
    original_image,
    perturbed_image,
    original_heatmap,
    perturbed_heatmap,
    save_path,
    title
):
    original_overlay = create_overlay(original_image, original_heatmap)
    perturbed_overlay = create_overlay(perturbed_image, perturbed_heatmap)
    plt.figure(figsize=(16, 8))
    plt.subplot(2, 3, 1)
    plt.imshow(original_image)
    plt.axis("off")
    plt.title("Original image")

    plt.subplot(2, 3, 2)
    plt.imshow(original_heatmap, cmap="hot")
    plt.axis("off")
    plt.title("Original SHAP")

    plt.subplot(2, 3, 3)
    plt.imshow(original_overlay)
    plt.axis("off")
    plt.title("Original overlay")

    plt.subplot(2, 3, 4)
    plt.imshow(perturbed_image)
    plt.axis("off")
    plt.title("Perturbed image")

    plt.subplot(2, 3, 5)
    plt.imshow(perturbed_heatmap, cmap="hot")
    plt.axis("off")
    plt.title("Perturbed SHAP")

    plt.subplot(2, 3, 6)
    plt.imshow(perturbed_overlay)
    plt.axis("off")
    plt.title("Perturbed overlay")

    plt.suptitle(title)
    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    plt.close()
```

```python
# Load original SHAP outputs
original_shap_values = np.load(SHAP_VALUES_PATH)
original_shap_values = normalise_shap_return(original_shap_values)
meta_df = pd.read_csv(SHAP_METADATA_PATH)

print("Original SHAP values shape:", original_shap_values.shape)
print("Metadata shape:", meta_df.shape)

# Create SHAP background again
train_df = pd.read_csv(TRAIN_CSV)
train_dataset = TBXrayDataset(train_df, transform=base_transform)
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)
background_images = []
for images, labels, image_ids, image_paths in train_loader:
    background_images.append(images)
    if len(background_images) * BATCH_SIZE >= BACKGROUND_SIZE:
        break
background = torch.cat(background_images, dim=0)[:BACKGROUND_SIZE].to(DEVICE)
print("Background shape:", background.shape)

# Create SHAP GradientExplainer
print("Creating SHAP GradientExplainer...")
explainer = shap.GradientExplainer(
    wrapped_model,
    background
)

# Stability Experiment
PERTURBATIONS = [
    "brightness_plus",
    "brightness_minus",
    "contrast_plus",
    "contrast_minus",
    "gaussian_noise",
    "rotation"
]
results = []
for idx in tqdm(range(len(meta_df)), desc="Running stability tests"):
    row = meta_df.iloc[idx]
    image_id = row["image_id"]
    image_path = row["image_path"]
    true_label = int(row["true_label"])

    original_image_np = load_image_as_display_np(image_path)
    original_tensor = display_np_to_normalized_tensor(original_image_np)
    original_prob, original_pred = get_tb_probability(model, original_tensor)
    original_heatmap = shap_to_heatmap(original_shap_values[idx])
    for perturbation_type in PERTURBATIONS:
        perturbed_image_np = perturb_image(
            original_image_np,
            perturbation_type
        )
        perturbed_tensor = display_np_to_normalized_tensor(perturbed_image_np)
        perturbed_prob, perturbed_pred = get_tb_probability(model, perturbed_tensor)
        perturbed_batch = perturbed_tensor.unsqueeze(0).to(DEVICE)
        perturbed_shap = explainer.shap_values(
            perturbed_batch,
            nsamples=NSAMPLES,
            rseed=SEED
        )
        perturbed_shap = normalise_shap_return(perturbed_shap)
        perturbed_shap_single = perturbed_shap[0]
        perturbed_heatmap = shap_to_heatmap(perturbed_shap_single)

        cos_sim = cosine_similarity(original_heatmap, perturbed_heatmap)
        spear_sim = spearman_similarity(original_heatmap, perturbed_heatmap)
        jaccard_10 = jaccard_topk(original_heatmap, perturbed_heatmap, top_percent=10)

        probability_change = perturbed_prob - original_prob
        absolute_probability_change = abs(probability_change)

        prediction_changed = int(original_pred != perturbed_pred)

        results.append({
            "image_id": image_id,
            "image_path": image_path,
            "true_label": true_label,
            "perturbation": perturbation_type,
            "original_prediction": original_pred,
            "perturbed_prediction": perturbed_pred,
            "prediction_changed": prediction_changed,
            "original_tb_probability": original_prob,
            "perturbed_tb_probability": perturbed_prob,
            "probability_change": probability_change,
            "absolute_probability_change": absolute_probability_change,
            "cosine_similarity": cos_sim,
            "spearman_correlation": spear_sim,
            "jaccard_top10": jaccard_10
        })

        np.save(
            os.path.join(
                PERT_SHAP_DIR,
                f"{image_id}_{perturbation_type}_shap.npy"
            ),
            perturbed_shap_single
        )

        # Save example panels for the first few images
        if idx < 3:
            title = (
                f"{image_id} | {perturbation_type} | "
                f"Cos: {cos_sim:.3f} | Spearman: {spear_sim:.3f} | "
                f"Jaccard@10: {jaccard_10:.3f}"
            )
            save_path = os.path.join(
                EXAMPLE_DIR,
                f"{image_id}_{perturbation_type}_stability_panel.png"
            )
            save_stability_panel(
                original_image_np,
                perturbed_image_np,
                original_heatmap,
                perturbed_heatmap,
                save_path,
                title
            )


results_df = pd.DataFrame(results)
results_path = os.path.join(OUTPUT_DIR, "stability_results.csv")
results_df.to_csv(results_path, index=False)
print(f"Saved detailed stability results to {results_path}")

# Summary Results
summary_df = results_df.groupby("perturbation").agg(
    mean_cosine_similarity=("cosine_similarity", "mean"),
    std_cosine_similarity=("cosine_similarity", "std"),
    mean_spearman_correlation=("spearman_correlation", "mean"),
    std_spearman_correlation=("spearman_correlation", "std"),
    mean_jaccard_top10=("jaccard_top10", "mean"),
    std_jaccard_top10=("jaccard_top10", "std"),
    mean_absolute_probability_change=("absolute_probability_change", "mean"),
    std_absolute_probability_change=("absolute_probability_change", "std"),
    prediction_change_rate=("prediction_changed", "mean"),
    n=("image_id", "count")
).reset_index()

summary_path = os.path.join(OUTPUT_DIR, "stability_summary.csv")
summary_df.to_csv(summary_path, index=False)

print("\nStability summary:")
print(summary_df)

# Plot 1: Cosine Similarity by Perturbation
plt.figure(figsize=(10, 5))
plt.bar(
    summary_df["perturbation"],
    summary_df["mean_cosine_similarity"],
    yerr=summary_df["std_cosine_similarity"],
    capsize=4
)
plt.xticks(rotation=30, ha="right")
plt.ylabel("Mean cosine similarity")
plt.title("SHAP Stability: Cosine Similarity by Perturbation")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "stability_cosine_similarity.png"), dpi=300)
plt.close()

# Plot 2: Spearman Correlation by Perturbation
plt.figure(figsize=(10, 5))
plt.bar(
    summary_df["perturbation"],
    summary_df["mean_spearman_correlation"],
    yerr=summary_df["std_spearman_correlation"],
    capsize=4
)
plt.xticks(rotation=30, ha="right")
plt.ylabel("Mean Spearman correlation")
plt.title("SHAP Stability: Spearman Correlation by Perturbation")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "stability_spearman_correlation.png"), dpi=300)
plt.close()

# Plot 3: Jaccard Top-10 Similarity
plt.figure(figsize=(10, 5))
plt.bar(
    summary_df["perturbation"],
    summary_df["mean_jaccard_top10"],
    yerr=summary_df["std_jaccard_top10"],
    capsize=4
)
plt.xticks(rotation=30, ha="right")
plt.ylabel("Mean Jaccard similarity, top 10%")
plt.title("SHAP Stability: Top-10% Attribution Overlap")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "stability_jaccard_top10.png"), dpi=300)
plt.close()

# Plot 4: Probability Change
plt.figure(figsize=(10, 5))
plt.bar(
    summary_df["perturbation"],
    summary_df["mean_absolute_probability_change"],
    yerr=summary_df["std_absolute_probability_change"],
    capsize=4
)
plt.xticks(rotation=30, ha="right")
plt.ylabel("Mean absolute TB probability change")
plt.title("DenseNet121 Prediction Stability Under Perturbations")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "stability_probability_change.png"), dpi=300)
plt.close()

# Save settings
settings = {
    "seed": SEED,
    "image_size": IMG_SIZE,
    "background_size": BACKGROUND_SIZE,
    "nsamples": NSAMPLES,
    "model_path": MODEL_PATH,
    "shap_values_path": SHAP_VALUES_PATH,
    "shap_metadata_path": SHAP_METADATA_PATH,
    "perturbations": PERTURBATIONS,
    "metrics": [
        "cosine_similarity",
        "spearman_correlation",
        "jaccard_top10",
        "absolute_probability_change",
        "prediction_change_rate"
    ]
}

with open(os.path.join(OUTPUT_DIR, "stability_settings.json"), "w") as f:
    json.dump(settings, f, indent=4)
```
